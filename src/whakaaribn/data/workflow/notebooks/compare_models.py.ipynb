```python
import pandas as pd
from whakaaribn.visualize import trellis_plot
from whakaaribn import get_color, get_data
from whakaaribn.model import WhakaariModel
from whakaaribn.smile_model import WhakaariSmileModel
from whakaaribn.forecast import forecast
from aitana import whakaari
from whakaaribn.util import assign_group_labels
```

```python
data = pd.read_csv(get_data("data/whakaaribn_data_release_v2.2.csv"), parse_dates=True, index_col=0)
data.index = data.index.tz_localize("utc")
eruptions = whakaari.eruptions(2, "0D", end_date=data.index[-1])
data_with_groups = assign_group_labels(data, eruptions, startdate=data.index[0], enddate=data.index[-1], ndays=30, min_interval=360)
data = data_with_groups
data
```

```python
bins = [0, 5, 20, 80, 95, 100]
pew = 40
smile_forecast = forecast(data, pew=pew, bins=bins, model_class=WhakaariSmileModel,
                          smoothing=30)
pgm_forecast = forecast(data, pew=pew, bins=bins, model_class=WhakaariModel,
                        smoothing=30)
```

```python
models = {'pgmpy': {'model': pgm_forecast['probs'], 'color': get_color(0)},
          'smile': {'model': smile_forecast['probs'], 'color': get_color(1), 'dash': 'dash'}}
```

```python
trellis_plot(models, data)
```
